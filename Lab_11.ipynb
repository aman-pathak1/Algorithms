# Travelling Salesman Problem using Dynamic Programming.
Problem Description:
Given a set of cities and distances, find the shortest possible route that visits each city once and returns to the starting city.

#Theory
TSP uses dynamic programming with bit masking. Store results of visited subsets to reduce repeated work.

## Algorithm
1. Start from source city
2. Mark visited cities using bit mask
3. Try all unvisited cities
4. Choose minimum cost path
5. Return to source city

```python
def tsp(graph):
    n = len(graph)
    visited = [False] * n
    min_cost = float('inf')

    def dfs(city, count, cost):
        nonlocal min_cost

        # agar sab cities visit ho gayi
        if count == n and graph[city][0] > 0:
            min_cost = min(min_cost, cost + graph[city][0])
            return

        # next city try karo
        for i in range(n):
            if not visited[i] and graph[city][i] > 0:
                visited[i] = True
                dfs(i, count + 1, cost + graph[city][i])
                visited[i] = False   # backtrack

    visited[0] = True
    dfs(0, 1, 0)

    return min_cost
```

```python
graph = [
[0, 10, 15, 20],
[10, 0, 35, 25],
[15, 35, 0, 30],
[20, 25, 30, 0]
]

print("Minimum travelling cost:", tsp(graph))
```
